```python
from deepface import DeepFace

models_name = ["VGG-Face", "Facenet", "Facenet512", "OpenFace",
               "DeepFace", "DeepID", "ArcFace", "Dlib", "SFace", 'Ensemble']

# 使用稳定的模型
model_name = models_name[2]  # Facenet512

# 图像对列�?- 每一对用于对比
img_pairs = [
    {"pair_name": "测试�?", "img1": "images/333.jpg", "img2": "images/222.jpg"},
    {"pair_name": "测试�?", "img1": "images/pengyuyan.jpg", "img2": "images/wuyanzu.webp"}
]

# 度量标准
metrics = ["cosine", "euclidean", "euclidean_l2"]

print(f"使用模型: {model_name}")
print(f"距离度量: {metrics[0]}\n")

# 批量处理多对图像
results = []
for pair in img_pairs:
    print(f"正在处理: {pair['pair_name']}")
    try:
        result = DeepFace.verify(
            img1_path=pair['img1'],
            img2_path=pair['img2'],
            model_name=model_name,
            distance_metric=metrics[0]  # cosine
        )
        results.append({
            'pair_name': pair['pair_name'],
            'verified': result['verified'],
            'distance': result['distance'],
            'threshold': result['threshold']
        })
        print(f"  结果: {'�?同一�? if result['verified'] else '�?不同�?}")
        print(f"  距离: {result['distance']:.4f} (阈�? {result['threshold']})\n")
    except Exception as e:
        print(f"  错误: {str(e)}\n")
        results.append({
            'pair_name': pair['pair_name'],
            'error': str(e)
        })

# 展示汇总结果
print("\n" + "="*50)
print("汇总结�?")
print("="*50)
for i, result in enumerate(results, 1):
    print(f"{i}. {result['pair_name']}")
    if 'error' in result:
        print(f"   状�? 处理失败 - {result['error']}")
    else:
        print(f"   验证: {'通过 �? if result['verified'] else '未通过 �?}")
        print(f"   距离: {result['distance']:.4f}")
    print()
```
